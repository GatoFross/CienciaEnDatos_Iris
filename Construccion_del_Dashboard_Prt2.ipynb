```python
# =========================

# ESTILO KPI CARD

# =========================

kpi_style = {

  "backgroundColor": "#ffffff",

  "padding": "20px",

  "borderRadius": "12px",

  "boxShadow": "2px 2px 12px rgba(0,0,0,0.1)",

  "textAlign": "center",

  "margin": "10px"

}



# =========================

# CALLBACK

# =========================

@callback(

  Output("grafico-hist", "figure"),

  Output("grafico-box", "figure"),

  Output("grafico-pie", "figure"),

  Output("kpi-container", "children"),

  Input("radio-variable", "value"),

  Input("filtro-especie", "value")

)

def actualizar_dashboard(variable, especie):



  dff = df.copy()



  if especie:

    dff = dff[dff["species"] == especie]



  # ================= HISTOGRAMA =================

  hist = px.histogram(

    dff,

    x="species",

    y=variable,

    color="species",

    histfunc="avg",

    title=f"Promedio de {variable} por especie"

  )

  hist.update_layout(template="plotly_white")



  # ================= BOXPLOT =================

  box = px.box(

    dff,

    x="species",

    y=variable,

    color="species",

    title=f"Distribución de {variable}"

  )

  box.update_layout(template="plotly_white")



  # ================= PIE =================

  pie = px.pie(

    dff,

    names="species",

    title="Distribución de especies"

  )

  pie.update_traces(textinfo="percent+label")



  # ================= KPIs =================

  kpi1 = html.Div([

    html.H4("📊 Total registros"),

    html.H2(len(dff))

  ], style=kpi_style)



  kpi2 = html.Div([

    html.H4("📈 Promedio"),

    html.H2(round(dff[variable].mean(), 2))

  ], style=kpi_style)



  kpi3 = html.Div([

    html.H4("📌 Máximo"),

    html.H2(round(dff[variable].max(), 2))

  ], style=kpi_style)



  kpi4 = html.Div([

    html.H4("📉 Mínimo"),

    html.H2(round(dff[variable].min(), 2))

  ], style=kpi_style)



  kpi_row = html.Div(

    className="row",

    children=[

      html.Div(kpi1, className="three columns"),

      html.Div(kpi2, className="three columns"),

      html.Div(kpi3, className="three columns"),

      html.Div(kpi4, className="three columns"),

    ],

    style={"marginBottom": "20px"}

  )



  return hist, box, pie, kpi_row





# =========================

# RUN APP

# =========================

if __name__ == "__main__":

  app.run(debug=True, port=8005, jupyter_mode="external")
```
